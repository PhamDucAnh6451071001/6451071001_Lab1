```python
#Câu 1
import warnings
warnings.filterwarnings("ignore")

print("Phạm Đức Anh - 6451071001")

import time
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth

# Đọc file groceries.csv (thay đúng tên file bạn đã upload lên Colab)
transactions = []
with open('groceries.csv', 'r', encoding='utf-8') as f:
    for line in f:
        # Tách các mặt hàng trên mỗi dòng theo dấu phẩy và loại bỏ khoảng trắng thừa
        items = [item.strip() for item in line.strip().split(',') if item.strip() != '']
        if items:
            transactions.append(items)

print("Tổng số giao dịch đọc được:", len(transactions))
print("5 giao dịch đầu tiên:")
for i in range(5):
    print(f"Giao dịch {i+1}: {transactions[i]}")
```

```python
#Câu 2
print("Phạm Đức Anh - 6451071001")

te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_groceries = pd.DataFrame(te_ary, columns=te.columns_)

print("Kích thước ma trận One-hot encoding (Số giao dịch x Số mặt hàng):", df_groceries.shape)
df_groceries.head(10)
```

```python
#Câu 3 - a
print("Phạm Đức Anh - 6451071001")

# Các ngưỡng min_sup khảo sát (từ 1% đến 10%)
min_sups = [0.01, 0.02, 0.03, 0.05, 0.07, 0.1]
apriori_times = []
apriori_counts = []

print("--- KẾT QUẢ GIẢI THUẬT APRIORI ---")
for sup in min_sups:
    start_time = time.time()
    itemsets_ap = apriori(df_groceries, min_support=sup, use_colnames=True)
    elapsed_time = time.time() - start_time

    apriori_times.append(elapsed_time)
    apriori_counts.append(len(itemsets_ap))
    print(f"min_sup = {sup*100:4.1f}% ({sup:.2f}) | Số tập phổ biến: {len(itemsets_ap):4d} | Thời gian: {elapsed_time:.4f} giây")
```

```python
#Câu 4 - b
print("Phạm Đức Anh - 6451071001")

fpgrowth_times = []
fpgrowth_counts = []

print("--- KẾT QUẢ GIẢI THUẬT FP-GROWTH ---")
for sup in min_sups:
    start_time = time.time()
    itemsets_fp = fpgrowth(df_groceries, min_support=sup, use_colnames=True)
    elapsed_time = time.time() - start_time

    fpgrowth_times.append(elapsed_time)
    fpgrowth_counts.append(len(itemsets_fp))
    print(f"min_sup = {sup*100:4.1f}% ({sup:.2f}) | Số tập phổ biến: {len(itemsets_fp):4d} | Thời gian: {elapsed_time:.4f} giây")
```

```python
#Câu 5
print("Phạm Đức Anh - 6451071001")

df_compare = pd.DataFrame({
    'min_sup': min_sups,
    'Số tập phổ biến': apriori_counts,
    'Thời gian Apriori (giây)': [round(t, 4) for t in apriori_times],
    'Thời gian FP-Growth (giây)': [round(t, 4) for t in fpgrowth_times]
})

print("--- BẢNG SO SÁNH THỜI GIAN XỬ LÝ GIỮA APRIORI VÀ FP-GROWTH ---")
df_compare
```

```python
#Câu 6 - c
print("Phạm Đức Anh - 6451071001")

plt.figure(figsize=(9, 5))
plt.plot(min_sups, apriori_times, marker='o', linewidth=2, color='crimson', label='Apriori')
plt.plot(min_sups, fpgrowth_times, marker='s', linewidth=2, color='dodgerblue', label='FP-Growth')

plt.title('So sánh thời gian xử lý giữa Apriori và FP-Growth (Groceries Dataset)')
plt.xlabel('Ngưỡng hỗ trợ tối thiểu (min_sup)')
plt.ylabel('Thời gian xử lý (giây)')
plt.xticks(min_sups, [f"{s*100:.0f}%" for s in min_sups])
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
